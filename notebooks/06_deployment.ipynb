# Phase 6: Deployment

The goal of this phase is to describe how the selected **Tuned Gradient Boosting** model could be integrated into the company’s marketing operations.

The model would generate response probabilities for customers before a new campaign. These predictions could help the Marketing Department prioritise customers who are more likely to respond and reduce unnecessary campaign contacts.


## 6.1 Plan Deployment

The proposed deployment process consists of the following steps:

### 1. Prepare New Customer Data

New customer data must undergo the same preparation steps used during model development, including:

* handling missing values;
* creating the engineered features `Age`, `Seniority`, `Total_Spending`, and `Accepted_Total`;
* encoding categorical variables;
* ensuring that the predictor columns match those used during training.

### 2. Generate Response Probabilities

The prepared data would be provided to the Tuned Gradient Boosting model.

For each customer, the model would generate:

* a predicted response class;
* a probability of responding to the campaign.

### 3. Create the Campaign Targeting List

Customers could be ranked according to their predicted response probabilities.

The resulting targeting list could include:

* customer identification;
* predicted response probability;
* predicted response class;
* campaign-priority category.

Customer identifiers should be retained separately for operational use and should not be included as predictors in the model.

### 4. Deliver the Results

The Marketing Department could access the predictions through a dashboard, report, or exported customer list.

## 6.2 Plan Monitoring and Maintenance

After each campaign, the model’s predictions should be compared with the actual customer responses.

The following metrics should be monitored:

* Precision;
* Recall;
* F1-score;
* ROC-AUC;
* false positives;
* false negatives.

The final test-set results provide the initial performance reference:

* Accuracy: **0.90**
* Precision: **0.73**
* Recall: **0.45**
* F1-score: **0.56**
* ROC-AUC: **0.90**

Data-quality checks should also confirm that:

* all required variables are available;
* missing values are handled consistently;
* categorical variables use the expected categories;
* engineered features are calculated using the same definitions;
* input columns match the columns used during training.

The model should be reviewed when performance declines or when customer behaviour and campaign conditions change considerably.

## 6.3 Produce Final Report

The project developed and compared several classification models for predicting customer response to a marketing campaign.

Based on the final test-set comparison, **Tuned Gradient Boosting** was selected as the final model.

Its results were:

* Accuracy: **0.90**
* Precision: **0.73**
* Recall: **0.45**
* F1-score: **0.56**
* ROC-AUC: **0.90**
* True negatives: **370**
* False positives: **11**
* False negatives: **36**
* True positives: **30**

The model classified 41 customers as likely responders. Of these, 30 actually responded.

Tuned Logistic Regression achieved the highest Recall, while the baseline Gradient Boosting model achieved the highest Precision. Tuned Gradient Boosting was selected because it achieved the highest overall F1-score and ROC-AUC.

The final report is intended for the Marketing Department and other stakeholders responsible for campaign planning and customer targeting.

## 6.4 Final Project Review

The project followed the CRISP-DM phases from business understanding through deployment planning.

The main outcomes were:

* customer data was cleaned and prepared for classification;
* class imbalance was examined and addressed during model development;
* Logistic Regression, Decision Tree, Random Forest, and Gradient Boosting models were compared;
* hyperparameter tuning was performed using GridSearchCV;
* Tuned Gradient Boosting was selected as the final model.

The selected model correctly identified **30 of the 66 responders**, produced **11 false positives**, and achieved an F1-score of **0.56**.

A limitation of the project is that the dataset does not include the financial cost of contacting a customer or the value of a successful response.

Future improvements could include:

* selecting the classification threshold using campaign costs and expected revenue;
* retraining with newly collected campaign data;
* developing a dashboard or automated targeting report.
