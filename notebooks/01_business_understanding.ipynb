# Marketing Campaign Response Prediction

# 1. Business Understanding

## 1.1 Business context

The company is a retail business that runs marketing campaigns to promote its products and increase customer purchases. Marketing campaigns can be expensive when they are sent to all customers without distinction. Some customers are more likely to respond positively than others, depending on their profile, purchase history, income, family situation, and previous interactions with the company.

The Marketing Department wants to improve the efficiency of future campaigns by identifying customers who are most likely to accept an offer.

## 1.2 Business objective

The main business objective is to support the Marketing Department in targeting the right customers for future campaigns.

By predicting which customers are likely to respond, the company can:

- reduce marketing costs by avoiding contacting customers with a low probability of response;
- increase the campaign response rate;
- improve return on investment;
- personalize marketing actions based on customer characteristics;
- focus sales and communication efforts on high-potential customers.

## 1.3 Machine learning objective

From a machine learning perspective, this project is a binary classification problem.

The target variable is `Response`.

- `Response = 1`: the customer accepted the offer in the last campaign.
- `Response = 0`: the customer did not accept the offer in the last campaign.

The goal is to build a predictive model that estimates whether a customer is likely to respond positively to a marketing campaign based on historical customer data.

## 1.4 Expected output

The expected output of the model is a prediction for each customer:

- class `1`: likely to respond to the campaign;
- class `0`: unlikely to respond to the campaign.

The model may also provide a probability score, allowing the Marketing Department to rank customers from the most likely to the least likely to respond.

## 1.5 Success criteria

The model should not only achieve good predictive performance, but also be useful from a business perspective.

Since the company wants to identify customers likely to respond, important evaluation metrics include:

- Recall: to detect as many potential responders as possible;
- Precision: to avoid targeting too many customers who will not respond;
- F1-score: to balance precision and recall;
- ROC-AUC: to evaluate the model’s ability to separate responders from non-responders.

Accuracy alone may not be sufficient, especially if the dataset is imbalanced.

## 1.6 Initial assumptions

Before exploring the data, we can make several assumptions:

- Customers who accepted previous campaigns may be more likely to accept the last campaign.
- Customers with higher spending amounts may be more responsive to marketing offers.
- Income may influence the probability of response.
- Recency may be important: customers who purchased recently may be more engaged.
- The number of web visits and purchases may reflect customer interest.
- Family situation, such as the number of children or teenagers at home, may influence purchasing behavior.
- Education and marital status may also have an impact on campaign response.
