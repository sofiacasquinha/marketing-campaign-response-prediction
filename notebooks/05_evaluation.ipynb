## 5.1 Evaluate Results

### 5.1.1 Evaluation criteria from Business Understanding

The main objective of this project was to develop a tool capable of supporting better marketing decisions. From a business perspective, a useful model should help the company identify customers who are more likely to respond to future campaigns and reduce resources spent on customers with a very low probability of engagement.

For this reason, model evaluation was based not only on statistical performance, but also on its potential business value.

The following criteria were considered:

- **Ability to identify likely responders:** The model should successfully detect customers with a higher probability of accepting campaign offers. Correctly identifying these customers allows the Marketing Department to focus its efforts on individuals with greater conversion potential.

- **Reduction of inefficient targeting:** Contacting the entire customer base can generate unnecessary costs and reduce campaign efficiency. The model should help prioritize customers and reduce communication directed toward individuals unlikely to respond.

- **Support for marketing decisions:** In addition to generating predictions, the model should provide useful information for campaign planning and customer targeting strategies.

- **Balance between Recall and Precision:** Since the dataset showed strong class imbalance, particular attention was given to these metrics. Recall measures the model’s ability to detect customers who actually respond, while Precision measures how reliable those positive predictions are. A model with very high Recall but very low Precision may target many customers unnecessarily, while a model with high Precision but low Recall may miss valuable opportunities.

- **ROC-AUC:** ROC-AUC evaluates how well the model separates responders from non-responders across different probability thresholds. A higher value indicates stronger class discrimination and greater flexibility for future business decisions.

Since the dataset was imbalanced, Accuracy alone was not enough. The focus was on finding a model that could support useful marketing decisions while keeping a good balance across the main metrics.

#### Key Findings

The models produced different results across the evaluation metrics.

**Tuned Gradient Boosting** achieved the highest test-set F1-score and ROC-AUC:

* Accuracy: **0.90**
* Precision: **0.73**
* Recall: **0.45**
* F1-score: **0.56**
* ROC-AUC: **0.90**

The model correctly identified **30 responders**, produced **11 false positives**, and missed **36 responders**.

**Tuned Logistic Regression** achieved the highest Recall at **0.74**, correctly identifying **49 responders**. However, it also produced **61 false positives**.

**Gradient Boosting** achieved the highest Precision at **0.78**, but its Recall was lower at **0.38**.

**Random Forest** achieved Recall of **0.61** and an F1-score of **0.54**.

The Decision Tree models produced the lowest F1-scores among the tested approaches.

### Final Model Selection: Tuned Gradient Boosting

### 5.1.4 Business Interpretation of the Selected Model

The selected Tuned Gradient Boosting model:

* correctly identified **30 customers** who responded to the campaign;
* missed **36 customers** who responded;
* incorrectly selected **11 customers** who did not respond;
* correctly excluded **370 customers** who did not respond.

The model classified 41 customers as likely responders, of whom 30 actually responded.

These results show that the model limited the number of unnecessary positive predictions while identifying a portion of the responder group.


### 5.1.5. Findings Beyond the Model

The analysis identified several customer characteristics associated with campaign response:

* customers with more recent purchases tended to show higher response rates;
* previous campaign acceptance was associated with future response;
* higher customer spending was associated with greater campaign engagement;
* customers with more catalogue and web purchases tended to respond more frequently;
* household composition showed differences in response behaviour.

These findings are based on patterns observed in the available dataset and do not establish causal relationships.



## 5.2. Process Review

## 5.2 Process Review

The project followed the main CRISP-DM phases: Business Understanding, Data Understanding, Data Preparation, Modeling, Evaluation, and Deployment planning.

Data-quality issues were identified and corrected before modeling. The variables `Age`, `Seniority`, `Total_Spending`, and `Accepted_Total` were created to represent customer characteristics and behaviour.

Missing values were handled after the train-test split. Scaling was applied to Logistic Regression, while tree-based models used non-scaled data. SMOTE was applied only to the training data used for the baseline models. Four classification algorithms were tested, and hyperparameter tuning was performed using GridSearchCV. Tuning improved some models but did not improve every evaluation metric in all cases.

The main limitation was the absence of information about the actual cost of contacting a customer and the value of a successful response. Therefore, model selection was based only on statistical metrics. The models were also evaluated using one historical train-test split. Performance may differ when the model is applied to future campaigns or new customer data.
